```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_theme(style="whitegrid")

роки_числа = [1950, 1960, 1970, 1990, 2000, 2012, 2014, 2019]

# Модель народжуваності України взагалом
plt.figure(figsize=(10, 5))
plt.plot(роки_числа, df.loc['Україна', роки], marker='o', color='royalblue', linewidth=2.5)
plt.title('Графік 1: Динаміка коефіцієнта народжуваності в Україні (1950–2019)', fontsize=14, fontweight='bold')
plt.xlabel('Рік', fontsize=12)
plt.ylabel('Кількість народжених на 1000 осіб', fontsize=12)
plt.xticks(роки_числа)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

#Порівняння контрастних регіонів
plt.figure(figsize=(10, 5))
обрані_регіони = ['Київ', 'Закарпатська', 'Чернігівська']
кольори = ['#e74c3c', '#2ecc71', '#f1c40f']

for рег, кол in zip(обрані_регіони, кольори):
    plt.plot(роки_числа, df.loc[рег, роки], marker='s', label=рег, color=кол, linewidth=2)

plt.title('Графік 2: Порівняння динаміки народжуваності в окремих регіонах', fontsize=14, fontweight='bold')
plt.xlabel('Рік', fontsize=12)
plt.ylabel('Коефіцієнт на 1000 осіб', fontsize=12)
plt.xticks(роки_числа)
plt.legend(title="Регіони")
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

# Мін / Макс / Середнє
plt.figure(figsize=(10, 5))
мін_знач = тільки_регіони[роки].min()
макс_знач = тільки_регіони[роки].max()
середні_знач = тільки_регіони[роки].mean()

plt.fill_between(роки_числа, мін_знач,  макс_знач, color='lightblue', alpha=0.4, label='Коридор розбіжності регіонів')
plt.plot(роки_числа, середні_знач, color='navy', linestyle='--', marker='^', label='Середнє по регіонах')

plt.title('Графік 3: Діаграма розподілу та регіонального коридору народжуваності', fontsize=14, fontweight='bold')
plt.xlabel('Рік', fontsize=12)
plt.ylabel('Коефіцієнт на 1000 осіб', fontsize=12)
plt.xticks(роки_числа)
plt.legend()
plt.show()

#Топ-5 регіонів у 2019 році
plt.figure(figsize=(10, 5))
топ5_2019 = тільки_регіони['2019'].dropna().sort_values(ascending=False).head(5)
sns.barplot(x=топ5_2019.values, y=топ5_2019.index, palette='viridis')
plt.title('Graphic 4: Топ-5 регіонів з найвищою народжуваністю у 2019 році', fontsize=14, fontweight='bold')
plt.xlabel('Коефіцієнт (на 1000 осіб)', fontsize=12)
plt.ylabel('Регіон', fontsize=12)
plt.show()

# Порівняння 1950 проти 2019
plt.figure(figsize=(12, 6))
регіони_порівняння = ['Волинська', 'Дніпропетровська', 'Одеська', 'Сумська', 'Київ']
df_порівняння = df.loc[регіони_порівняння, ['1950', '2019']].reset_index()
df_трансформований = pd.melt(df_порівняння, id_vars='Регіон', var_name='Рік', value_name='Коефіцієнт')

sns.barplot(x='Регіон', y='Коефіцієнт', hue='Рік', data=df_трансформований, palette='muted')
plt.title('Графік 5: Порівняльний аналіз народжуваності (1950 проти 2019 року)', fontsize=14, fontweight='bold')
plt.xlabel('Регіон', fontsize=12)
plt.ylabel('Коефіцієнт на 1000 осіб', fontsize=12)
plt.show()
```
