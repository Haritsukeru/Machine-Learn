```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('bestsellers with categories.csv')

print(df.head())

df.info()
```

```python
print(df.isnull().sum())

df_clean = df.drop_duplicates(subset=['Name', 'Author'])
print(len(df))
print(f"Тільки унікальні книги: {len(df_clean)}")
```

```python
print("Топ-5 найпопулярніших авторів")
print(df_clean['Author'].value_counts().head(5))

print("Порівння за показником ціни та рейтингу")
genre_comparison = df_clean.groupby('Genre')[['Price', 'User Rating']].mean()
print(genre_comparison)
```

```python
sns.set_theme(style="whitegrid")

# 1. Співвідношення жанрів серед унікальних книг
plt.figure(figsize=(6, 5))
sns.countplot(data=df_clean, x='Genre', palette='Set2')
plt.title('Кількість книг за жанрами (Fiction vs Non Fiction)')
plt.xlabel('Жанр')
plt.ylabel('Кількість книг')
plt.show()

# 2. Зміна середньої ціни книг по роках
plt.figure(figsize=(10, 5))
sns.lineplot(data=df, x='Year', y='Price', hue='Genre', marker='o', errorbar=None)
plt.title('Динаміка середньої ціни книг (2009 - 2019)')
plt.xlabel('Рік')
plt.ylabel('Середня ціна ($)')
plt.xticks(df['Year'].unique())
plt.show()

# 3. Вивід рейтингів користувачів
plt.figure(figsize=(8, 4))
sns.histplot(data=df_clean, x='User Rating', kde=True, bins=15, color='purple')
plt.title('Розподіл оцінок користувачів')
plt.xlabel('Рейтинг')
plt.ylabel('Частота')
plt.show()

# 4. Кореляція між ціною, рейтингом та кількістю відгуків
plt.figure(figsize=(6, 4))
corr_matrix = df_clean[['User Rating', 'Reviews', 'Price']].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Матриця кореляції числових ознак')
plt.show()
```
