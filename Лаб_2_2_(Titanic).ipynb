```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Зчитуємо дані
df = pd.read_csv('titanic.csv')

# Виводимо перші 5 рядків таблиці
df.head()
```

```python
print("Інформація про датасет")
df.info()

print("Опис")
df.describe()
```

```python
# Перевіряємо скільки пустих значень у кожному стовпці
print(df.isnull().sum())

# Наприклад, заповнимо пропущені значення віку медіанним значенням
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
# Налаштування стилю графіків
sns.set_theme(style="whitegrid")

# 1. Графік виживання
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Survived')
plt.title('Розподіл виживання (0 = Загинув, 1 = Вижив)')
plt.show()

# 2. Зв'язок між класом квитка та виживанням
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Pclass', hue='Survived')
plt.title('Виживання залежно від класу квитка')
plt.show()

# 3. Розподіл пасажирів за віком
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='Age', kde=True, bins=30)
plt.title('Розподіл пасажирів за віком')
plt.show()
```
